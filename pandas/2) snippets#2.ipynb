```python
import pandas as pd
```

```python
people={
    "first":["Abhishek","Ram","Shyam","Mohit"],
    "last":["Karn","Magar","Shah","Awasthi"],
    "email":["abhishek@gmail.com","rammagar@gmail.com","shyamah@yahoo.com","hitashi@gmail.com"],
    "Pass":['Yes',"No","Yes","Yes"],
    "Marks":[97,30,50,86]
}
```

```python
df=pd.DataFrame(people)
```

```python
df['Pass']=="Yes"
```

```python
filt=(df['Pass']=="Yes") & (df['Marks']>80)
```

```python
df[filt]
```

```python
df.loc[filt,'first']
```
